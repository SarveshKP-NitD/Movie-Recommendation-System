# Models Implemented

The project includes multiple recommendation approaches:

## 1. Popularity-Based Recommendation

Recommends the most popular movies based on average ratings and rating counts.

Advantages:

- Simple
- Fast
- No training required

Limitations:

- Same recommendation for every user

---

## 2. User-Based Collaborative Filtering

Finds users with similar preferences and recommends movies liked by similar users.

Advantages:

- Personalized recommendations

Limitations:

- Suffers from data sparsity

---

## 3. Item-Based Collaborative Filtering

Finds similar movies based on user rating behavior.

Advantages:

- Stable recommendations
- Widely used in production systems

---

## 4. Matrix Factorization (SVD)

Learns latent representations of users and movies.

Advantages:

- Better handling of sparse data
- Captures hidden relationships

---

## 5. FAISS Vector Search

Uses learned embeddings with Facebook AI Similarity Search (FAISS) for efficient nearest-neighbor search.

Advantages:

- Extremely fast retrieval
- Scalable to large datasets

---

## 6. Neural Collaborative Filtering (NCF)

Uses deep neural networks to learn complex user-item interactions.

Advantages:

- High recommendation accuracy
- Learns nonlinear relationships
- State-of-the-art recommendation approach

# Challenges Faced

During the development of this project, several practical challenges were encountered:

# Limitations

- No temporal information is used.
